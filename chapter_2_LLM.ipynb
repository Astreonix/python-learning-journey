```python
import urllib.request
url = ("https://raw.githubusercontent.com/rasbt/"
"LLMs-from-scratch/main/ch02/01_main-chapter-code/"
"the-verdict.txt")
file_path = "the-verdict.txt"
urllib.request.urlretrieve(url, file_path)
```

```python
with open ("the-verdict.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()
    print("the total number of character:", len(raw_text))
    print(raw_text[:99])
```

```python
import re
text = "hello, world. this is a text."
result = re.split(r'(\s)', text)
print(result)
```

```python
result = re.split(r'([,.]|\s)', text)
print(result)
```

```python
result = [item for item in result if item.strip()]
print(result)
```

```python
Mehr = "hello, world Is this-- a test?"
result = re.split(r'([,.:;?_!()\']|--|\s)', Mehr)
result = [item.strip() for item in result if item.strip()]
print(result)
```

```python
preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text)
preprocessed = [Mehr.strip() for Mehr in preprocessed if Mehr.strip()]
print(len(preprocessed))
```

```python
print(preprocessed[:30])
```

```python
all_words = sorted(set(preprocessed))  #sort and remove (set use for remove duplicate)
vocab_size = len(all_words)
print(vocab_size)

# conerting to numeric representation called token ID
```

```python
# creating  a vocabulary and enumarates (show all word index actually )
# use enumarates fun for the indexing
vocab = {token: integer for integer, token in enumerate(all_words)}
# every char has seprate token
for i, item in enumerate(vocab.items()):
    print(item)
    if i >= 50:
        break

```
