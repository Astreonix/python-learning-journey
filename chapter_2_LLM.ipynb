# now to convert toke id to text this process is also called as the decoding  
# on the above code i will covert text to token id this process is called encoding

```python
class SimpleTokenizerV1:
  def __init__(self, vocab):   # -- use double this for the syntax
    self.str_to_int = vocab
    self.int_to_str = {i:s for s,i in vocab.items() }

    # define both encode and decode function in the tokenizer class

  def encode(self, text) :
    preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
    preprocessed = [
        Mehr.strip() for Mehr in preprocessed if Mehr.strip()
        ]
    ids = [self.str_to_int[word] for word in preprocessed]
    return ids

  def decode(self, ids):
    text = " ".join([self.int_to_str[i] for i in ids])

    text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
    return text
```

```python
tokenizer = SimpleTokenizerV1(vocab)   # use the vocabulary to encoding
    # encode function inn the tokenizer class
text = """"It's the last he painted, you know,"
Mrs. Gisburn said with pardonable pride."""
ids = tokenizer.encode(text)
print(ids)
```

```python
print(tokenizer.decode(ids))   # decode fun in the tokenizer class
```

# A simple text tokenzier that handles unknown words :

```python


# Redefining SimpleTokenizerV2 with correct indentation
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i: s for s, i in vocab.items()}

    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
        preprocessed = [
            item.strip() for item in preprocessed if item.strip()
        ]
        preprocessed = [
            item if item in self.str_to_int else "<|unk|>" for item in preprocessed
        ]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids

    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        text = re.sub(r'\s+([,.:;?!"()\'])', r'\1', text)
        return text
```

```python
text1 = "Hello, do you like tea?"
text2 = "In the sunlit terraces of the palace."
text = " <|endoftext|> ".join((text1, text2))
print(text)
```

```python

tokenizer = SimpleTokenizerV2(vocab)
print(tokenizer.encode(text))
```

```python
print(tokenizer.decode(tokenizer.encode(text)))
```

```python
from importlib_metadata import version
import tiktoken
print("tiktoken version:", version("tiktoken"))
```

```python
tokenizer = tiktoken.get_encoding("gpt2")
```

```python
text = (
    "Hello, do you like tea?  <|endoftext|> In the sunlit terraces"
      "of someunknownplace."
)
integers = tokenizer.encode(text, allowed_special={"<|endoftext|>"})
print(integers) # this code part will print the token ids because of the encode
```

```python
strings = tokenizer.decode(integers)
print(strings)   # this part can convert token id to text bcz of the decode
```
